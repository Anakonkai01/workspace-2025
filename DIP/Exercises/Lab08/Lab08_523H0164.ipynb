```python
import cv2
from matplotlib import pyplot

STUDENT_ID = '523H0164'

cascade_path = 'haarcascade_frontalface_default.xml'
image_input_path = 'test1.jpg'
image_output_path = 'image_10.1_01.png'

face_cascade = cv2.CascadeClassifier(cascade_path)
pixels_rgb = pyplot.imread(image_input_path)
image_bgr = cv2.cvtColor(pixels_rgb, cv2.COLOR_RGB2BGR)
gray = cv2.cvtColor(image_bgr, cv2.COLOR_BGR2GRAY)

faces = face_cascade.detectMultiScale(gray, 1.1, 4)
print(f"Found {len(faces)} faces.")

for (x, y, w, h) in faces:
    cv2.rectangle(image_bgr, (x, y), (x+w, y+h), (255, 0, 0), 2)

cv2.putText(image_bgr, STUDENT_ID, (5, 25), cv2.FONT_HERSHEY_SIMPLEX, 1, (0, 0, 255), 2)

cv2.imwrite(image_output_path, image_bgr)
print(f"Saved result to '{image_output_path}'")

cv2.imshow("image",image_bgr)
```

```python
import cv2
from matplotlib import pyplot
from mtcnn.mtcnn import MTCNN
from google.colab.patches import cv2_imshow

STUDENT_ID = '523H0164'

image_input_path = 'test1.jpg'
image_output_path = 'image_10.1_02.png'

pixels_rgb = pyplot.imread(image_input_path)
detector = MTCNN()
results = detector.detect_faces(pixels_rgb)

print(f"Found {len(results)} faces.")

image_bgr = cv2.cvtColor(pixels_rgb, cv2.COLOR_RGB2BGR)

for face in results:
    x, y, w, h = face['box']
    cv2.rectangle(image_bgr, (x, y), (x+w, y+h), (0, 255, 0), 2)
    
    keypoints = face['keypoints']
    for point in ['left_eye', 'right_eye', 'nose', 'mouth_left', 'mouth_right']:
        cv2.circle(image_bgr, keypoints[point], 2, (0, 0, 255), 2)

cv2.putText(image_bgr, STUDENT_ID, (5, 25), cv2.FONT_HERSHEY_SIMPLEX, 1, (0, 0, 255), 2)

cv2.imwrite(image_output_path, image_bgr)
print(f"Saved result to '{image_output_path}'")

cv2.imshow("image",image_bgr)
```

```python
import cv2
import matplotlib.pyplot as plt
from pyzbar import pyzbar
import numpy as np

STUDENT_ID = "523H0164"

def add_text_bottom_right(text, img, fontFace=cv2.FONT_HERSHEY_SIMPLEX,
                          fontScale=1.0, color=(0, 255, 0), thickness=2, margin=20):
    image_with_text = img.copy()
    text_size, _ = cv2.getTextSize(text, fontFace, fontScale, thickness)
    text_width, text_height = text_size

    org_x = max(0, image_with_text.shape[1] - text_width - margin)
    org_y = max(text_height, image_with_text.shape[0] - margin)

    cv2.putText(image_with_text, text, (org_x, org_y), fontFace, fontScale,
                color, thickness, cv2.LINE_AA)
    return image_with_text


def analyze_barcode_lines(img, barcode_rect):
    """Extract and analyze individual lines in a barcode"""
    x, y, w, h = barcode_rect
    
    # Extract barcode region with some padding
    padding = 5
    x1 = max(0, x - padding)
    y1 = max(0, y - padding)
    x2 = min(img.shape[1], x + w + padding)
    y2 = min(img.shape[0], y + h + padding)
    
    barcode_region = img[y1:y2, x1:x2]
    gray_region = cv2.cvtColor(barcode_region, cv2.COLOR_BGR2GRAY)
    
    # Take a horizontal slice through the middle of the barcode
    mid_height = gray_region.shape[0] // 2
    scan_line = gray_region[mid_height, :]
    
    # Threshold to binary
    _, binary_line = cv2.threshold(scan_line, 0, 255, cv2.THRESH_BINARY + cv2.THRESH_OTSU)
    
    # Find transitions (edges between bars and spaces)
    transitions = []
    prev_value = binary_line[0]
    
    for i in range(1, len(binary_line)):
        if binary_line[i] != prev_value:
            transitions.append(i)
            prev_value = binary_line[i]
    
    # Calculate bar and space widths
    bars_info = []
    for i in range(len(transitions) - 1):
        width = transitions[i + 1] - transitions[i]
        is_bar = binary_line[transitions[i]] == 0  # Black is bar, white is space
        bars_info.append({
            'type': 'bar' if is_bar else 'space',
            'position': transitions[i] + x1,
            'width': width
        })
    
    return bars_info, scan_line, binary_line


img = cv2.imread('/content/barcode.jpg')

if img is None:
    print("Error: Cannot load image. Please check the path or file.")
else:
    rotations = [
        None,
        cv2.ROTATE_90_CLOCKWISE,
        cv2.ROTATE_180,
        cv2.ROTATE_90_COUNTERCLOCKWISE
    ]

    barcodes = []
    img_to_scan = img.copy()

    for rotation in rotations:
        if rotation is not None:
            img_to_scan = cv2.rotate(img, rotation)

        barcodes = pyzbar.decode(img_to_scan)

        if barcodes:
            print(f"Found barcode at rotation: {rotation}")
            break

    if barcodes:
        img_with_boxes = img.copy()

        for idx, barcode in enumerate(barcodes):
            barcode_data = barcode.data.decode('utf-8')
            barcode_type = barcode.type
            (x, y, w, h) = barcode.rect
            polygon_points = barcode.polygon

            print(f"\n{'='*60}")
            print(f"BARCODE #{idx + 1}")
            print(f"{'='*60}")
            print(f"Type: {barcode_type}")
            print(f"Data: {barcode_data}")
            print(f"Quality: {barcode.quality}")
            print(f"\nBounding Box:")
            print(f"  Position: x={x}, y={y}")
            print(f"  Size: width={w}, height={h}")
            print(f"  Area: {w * h} pixels")
            
            print(f"\nPolygon Points (corners):")
            for i, point in enumerate(polygon_points):
                print(f"  Point {i+1}: ({point.x}, {point.y})")
            
            # Analyze individual bars and spaces
            print(f"\nAnalyzing barcode structure...")
            bars_info, scan_line, binary_line = analyze_barcode_lines(img, barcode.rect)
            
            # Count bars and spaces
            bars = [b for b in bars_info if b['type'] == 'bar']
            spaces = [b for b in bars_info if b['type'] == 'space']
            
            print(f"  Total bars (black lines): {len(bars)}")
            print(f"  Total spaces (white gaps): {len(spaces)}")
            
            # Bar width statistics
            if bars:
                bar_widths = [b['width'] for b in bars]
                print(f"\nBar Width Statistics:")
                print(f"  Min width: {min(bar_widths)} pixels")
                print(f"  Max width: {max(bar_widths)} pixels")
                print(f"  Average width: {np.mean(bar_widths):.2f} pixels")
                print(f"  Median width: {np.median(bar_widths):.2f} pixels")
            
            # Show first 20 bars/spaces details
            print(f"\nFirst 20 bars and spaces details:")
            for i, element in enumerate(bars_info[:20]):
                print(f"  {i+1}. {element['type'].upper():6s} - Position: {element['position']:4d}, Width: {element['width']:2d} pixels")
            
            if len(bars_info) > 20:
                print(f"  ... and {len(bars_info) - 20} more elements")

            # Draw on image
            cv2.rectangle(img_with_boxes, (x, y), (x + w, y + h), (0, 255, 0), 2)
            cv2.putText(img_with_boxes, barcode_data, (x, y - 10),
                       cv2.FONT_HERSHEY_SIMPLEX, 0.5, (0, 255, 0), 2)

        img_with_boxes = add_text_bottom_right(STUDENT_ID, img_with_boxes,
                                               color=(255, 0, 255), fontScale=0.8,
                                               thickness=2, margin=15)

        # Display results
        fig, axes = plt.subplots(2, 1, figsize=(12, 10))
        
        img_rgb = cv2.cvtColor(img_with_boxes, cv2.COLOR_BGR2RGB)
        axes[0].imshow(img_rgb)
        axes[0].set_title("Image with Detected Barcodes")
        axes[0].axis('off')
        
        # Plot scan line profile
        if barcodes:
            barcode = barcodes[0]
            bars_info, scan_line, binary_line = analyze_barcode_lines(img, barcode.rect)
            axes[1].plot(scan_line, label='Gray scale intensity', linewidth=1)
            axes[1].plot(binary_line, label='Binary threshold', linewidth=1, alpha=0.7)
            axes[1].set_title("Barcode Scan Line Profile (horizontal slice through middle)")
            axes[1].set_xlabel("Position (pixels)")
            axes[1].set_ylabel("Intensity")
            axes[1].legend()
            axes[1].grid(True, alpha=0.3)
        
        plt.tight_layout()
        plt.show()

        cv2.imwrite("image_10.2_01.png", img_with_boxes)
        print(f"\n{'='*60}")
        print("Image saved as 'image_10.2_01.png'")
    else:
        print("No barcodes found at any rotation.")
```
