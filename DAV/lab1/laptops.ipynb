```python
import numpy as np
import pandas as pd
```

```python
laptop_data = pd.read_csv("laptops_asus_data_   cellphones_full_v2.csv")
```

```python
laptop_data.head()
```

```python
laptop_data.info(verbose=True)
```

```python
# make the Price, RAM, Storage, ScreenSize, ScreenResolution columns numeric
laptop_data['Price'] = pd.to_numeric(laptop_data['Price'], errors='coerce')
laptop_data['RAM'] = pd.to_numeric(laptop_data['RAM'], errors='coerce')
laptop_data['Storage'] = pd.to_numeric(laptop_data['Storage'], errors='coerce')
laptop_data['ScreenSize'] = pd.to_numeric(laptop_data['Screen Size'], errors='coerce')
laptop_data['ScreenResolution'] = pd.to_numeric(laptop_data['Screen Resolution'], errors='coerce')
```

```python
# Now we can run the info() method again to see the changes
laptop_data.info(verbose=True)
```
